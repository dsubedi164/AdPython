# Handling Missing, Categorical Data



## Handling Missing Data 

Missing data is common in datasets and can arise due to various reasons like data entry errors or non-responses. Handling it properly is crucial to avoid biased models.


### Techniques
1. **Deletion**: Remove rows/columns with missing values.
2. **Imputation**: Fill missing values with statistics (mean, median, mode) or advanced methods (KNN, interpolation).
3. **Prediction**: Use ML models to predict missing values.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
```

```python
# Create a sample dataset with missing values
data = {
    'Age': [25, np.nan, 35, 40, np.nan, 50],
    'Salary': [50000, 60000, np.nan, 80000, 90000, np.nan],
    'City': ['New York', 'Los Angeles', 'New York', np.nan, 'Chicago', 'Los Angeles']
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

### 1. Deletion Methods

```python
# Listwise deletion (drop rows with any missing values)
df_listwise = df.dropna()
print("\nAfter Listwise Deletion:")
print(df_listwise)

# Pairwise deletion (drop columns with missing values)
df_pairwise = df.dropna(axis=1)
print("\nAfter Pairwise Deletion:")
print(df_pairwise)
```

### 2. Simple Imputation

```python
# Mean imputation for numerical columns

# Calculate mean (ignores NaN automatically)
age_mean = df['Age'].mean()
salary_mean = df['Salary'].mean()

# Fill missing values
df['Age_mean'] = df['Age'].fillna(age_mean)
df['Salary_mean'] = df['Salary'].fillna(salary_mean)


# Mode imputation for categorical column

# Calculate mode
city_mode = df['City'].mode()[0]

# Fill missing values
df['City_mode'] = df['City'].fillna(city_mode)


print("\nAfter Mean and Mode Imputation:") 
print(df[['Age', 'Age_mean', 'Salary', 'Salary_mean', 'City', 'City_mode']])
```

## Handling Categorical Data
Categorical data represents categories or labels. Machine learning models require numerical inputs, so we encode them.

Categorical data is handled by converting categories into numerical form. Common methods include label encoding for ordinal data and one-hot encoding for nominal data. Missing categorical values are handled using mode imputation or by introducing a new category. Proper encoding is essential as most machine learning algorithms work only with numerical data.

### Types
- **Nominal**:  Nominal data consists of categories without any natural order or ranking. No order (e.g., colors).
- **Ordinal**: Ordinal data consists of categories that have a meaningful order or ranking, but the difference between ranks is not measurable. Ordered (e.g., sizes: small, medium, large).

### Techniques
1. **Label Encoding**: Label Encoding assigns a unique integer to each category without considering any order. Assign integers to categories means that each distinct category is replaced by a number.
Example for Label enoding : 


| Color | Encoded |
| ----- | ------- |
| Red   | 0       |
| Blue  | 1       |
| Green | 2       |

2. **One-Hot Encoding**: One-Hot Encoding converts each category into a separate binary column. Create binary columns for each category.
Example for one hot enoding :   

| City      | Kathmandu | Pokhara | Biratnagar |
| --------- | --------- | ------- | ---------- |
| Kathmandu | 1         | 0       | 0          |
| Pokhara   | 0         | 1       | 0          |

3. **Ordinal Encoding**: Ordinal Encoding assigns integers to categories based on their natural order or ranking. Assign integers based on order. 
Example for Ordinal Encoding : 


| Size   | Encoded |
| ------ | ------- |
| Small  | 1       |
| Medium | 2       |
| Large  | 3       |

4. **Target Encoding**: Target Encoding replaces categories with the mean of the target variable for that category. Use target variable to encode categories.
Example for Target Encoding :


| City      | Target (Salary) |
| --------- | --------------- |
| Kathmandu | 80,000          |
| Pokhara   | 60,000          |


```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
import category_encoders as ce
```

```python
# Sample dataset
data_cat = {
    'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red'],
    'Size': ['Small', 'Medium', 'Large', 'Medium', 'Small'],
    'Target': [1, 0, 1, 0, 1]
}
df_cat = pd.DataFrame(data_cat)
print("Original Categorical DataFrame:")
print(df_cat)
```

### 1. Label Encoding

```python
le = LabelEncoder()
df_cat['Color_label'] = le.fit_transform(df_cat['Color'])
print("\nAfter Label Encoding:")
print(df_cat[['Color', 'Color_label']])
```

### 2. One-Hot Encoding

```python
ohe = OneHotEncoder(sparse_output=False)
color_ohe = ohe.fit_transform(df_cat[['Color']])
color_ohe_df = pd.DataFrame(color_ohe, columns=ohe.get_feature_names_out(['Color']))
df_cat_ohe = pd.concat([df_cat, color_ohe_df], axis=1)
print("\nAfter One-Hot Encoding:")
print(df_cat_ohe)
```

### 3. Ordinal Encoding

```python
oe = OrdinalEncoder(categories=[['Small', 'Medium', 'Large' ]])
df_cat['Size_ordinal'] = oe.fit_transform(df_cat[['Size']])
print("\nAfter Ordinal Encoding:")
print(df_cat[['Size', 'Size_ordinal']])
```

### 4. Target Encoding

```python
te = ce.TargetEncoder(cols=['Color'],smoothing=0)
df_cat['Color_target'] = te.fit_transform(df_cat['Color'], df_cat['Target'])
print("\nAfter Target Encoding:")
print(df_cat[['Color', 'Color_target']])
```
