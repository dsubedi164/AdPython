# Probability Review, Sampling, and Hypothesis Testing

## Probability Review

### Basic Concepts
Probability is a measure of the likelihood that an event will occur. It ranges from 0 (impossible) to 1 (certain).

#### Axioms of Probability
1. For any event A, 0 ≤ P(A) ≤ 1.
2. P(Ω) = 1, where Ω is the sample space.
3. For mutually exclusive events A and B, P(A ∪ B) = P(A) + P(B).

### Conditional Probability
The probability of A given B is P(A|B) = P(A ∩ B) / P(B), if P(B) > 0.

### Independence
Events A and B are independent if P(A ∩ B) = P(A) * P(B).

### Bayes' Theorem
P(A|B) = [P(B|A) * P(A)] / P(B)

### Random Variables
A random variable X assigns a numerical value to each outcome in the sample space.
- Discrete: e.g., binomial, Poisson.
- Continuous: e.g., normal, exponential.

### Expected Value and Variance
E(X) = ∑ [x * P(X=x)] for discrete.
Var(X) = E(X²) - [E(X)]²

```python
import numpy as np
from scipy import stats

# Example: Basic Probability
# Probability of rolling a 3 or 5 on a fair die
prob = 2 / 6
print(f'Probability: {prob}')

# Conditional Probability Example
# P(A|B) where A: even number, B: number > 3
# Sample space: 1-6
# B: 4,5,6; A ∩ B: 4,6
p_a_given_b = 2/3
print(f'Conditional Probability: {p_a_given_b}')

# Bayes' Theorem Example
# P(Disease|Positive) = [P(Positive|Disease) * P(Disease)] / P(Positive)
p_disease = 0.01
p_pos_given_disease = 0.99
p_pos_given_no_disease = 0.05
p_positive = (p_pos_given_disease * p_disease) + (p_pos_given_no_disease * (1 - p_disease))
p_disease_given_pos = (p_pos_given_disease * p_disease) / p_positive
print(f'P(Disease|Positive): {p_disease_given_pos:.4f}')

# Random Variable: Binomial Distribution
# n=10, p=0.5, expected value = n*p
n, p = 10, 0.5
expected = n * p
variance = n * p * (1 - p)
print(f'Expected Value: {expected}, Variance: {variance}')
```

## Sampling

### Population vs. Sample
Population: Entire group of interest.
Sample: Subset of the population.

### Sampling Methods
- Simple Random Sampling: Each member has equal chance.
- Stratified Sampling: Divide into strata, sample from each.
- Cluster Sampling: Divide into clusters, sample clusters.
- Systematic Sampling: Select every kth member.

### Sampling Distribution
The distribution of a statistic (e.g., sample mean) over many samples.

### Central Limit Theorem (CLT)
For large sample sizes (n ≥ 30), the sampling distribution of the sample mean is approximately normal, regardless of the population distribution. Mean = population mean, SD = σ / √n (standard error).

### Bias and Variance in Sampling
Bias: Systematic error in sampling.
Variance: Variability due to random selection.

```python
# Example: Simple Random Sampling
population = np.arange(1, 101)  # Population 1 to 100
sample = np.random.choice(population, size=10, replace=False)
print(f'Sample: {sample}')
sample_mean = np.mean(sample)
print(f'Sample Mean: {sample_mean}')

# Demonstrating CLT
# Population: Uniform [0,1]
pop = np.random.uniform(0, 1, 10000)
pop_mean = np.mean(pop)
pop_std = np.std(pop)

# Take 1000 samples of size 30
sample_means = [np.mean(np.random.choice(pop, 30)) for _ in range(1000)]
sampling_mean = np.mean(sample_means)
sampling_std = np.std(sample_means)
theoretical_se = pop_std / np.sqrt(30)

print(f'Population Mean: {pop_mean:.4f}, Sampling Mean: {sampling_mean:.4f}')
print(f'Theoretical SE: {theoretical_se:.4f}, Observed SD: {sampling_std:.4f}')
```

## Hypothesis Testing

### Basics
Hypothesis testing is a method to make decisions using data. It involves:
- Null Hypothesis (H0): No effect or status quo.
- Alternative Hypothesis (H1): Effect or difference.

### Steps
1. State H0 and H1.
2. Choose significance level (α, e.g., 0.05).
3. Select test statistic (e.g., z, t).
4. Compute p-value or critical value.
5. Decide: Reject H0 if p-value < α.

### Types of Errors
- Type I: Reject true H0 (false positive).
- Type II: Fail to reject false H0 (false negative).

### Common Tests
- Z-test: Known variance, large n.
- T-test: Unknown variance.
- Chi-square: Categorical data.
- ANOVA: Compare means across groups.

### P-value
The probability of observing the data (or more extreme) assuming H0 is true.

```python
# Example: One-sample Z-test
# H0: mu = 100, sample mean = 105, sigma=15, n=30, alpha=0.05
mu0 = 100
x_bar = 105
sigma = 15
n = 30
alpha = 0.05

z = (x_bar - mu0) / (sigma / np.sqrt(n))
p_value = 1 - stats.norm.cdf(z)  # One-tailed
print(f'Z-statistic: {z:.4f}, P-value: {p_value:.4f}')
if p_value < alpha:
    print('Reject H0')
else:
    print('Fail to reject H0')

# One-sample T-test
data = np.array([98, 102, 105, 99, 101, 104, 100, 103])
t_stat, p_val = stats.ttest_1samp(data, popmean=100)
print(f'T-statistic: {t_stat:.4f}, P-value: {p_val:.4f}')

# Chi-square Test of Independence
contingency = np.array([[10, 20], [15, 25]])
chi2, p, dof, expected = stats.chi2_contingency(contingency)
print(f'Chi2: {chi2:.4f}, P-value: {p:.4f}')
```
